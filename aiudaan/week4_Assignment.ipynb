```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Datasets/dp_lms_student_analytics_dataset.csv")
df
```

```python

monthly_enrollments = df.groupby('Month').size()
monthly_revenue = df.groupby('Month')['Revenue'].sum()
monthly_progress = df.groupby('Month')['CourseProgress'].mean()
monthly_attendance = df.groupby('Month')['AttendancePercentage'].mean()

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly_enrollments.index.astype(str), monthly_enrollments.values, marker='o', color='steelblue')
ax.set_title('Monthly Student Enrollments')
ax.set_xlabel('Month')
ax.set_ylabel('Number of Students')
ax.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly_revenue.index.astype(str), monthly_revenue.values, marker='o', color='green')
ax.set_title('Monthly Revenue Trend')
ax.set_xlabel('Month')
ax.set_ylabel('Revenue')
ax.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly_progress.index.astype(str), monthly_progress.values, marker='o', color='orange')
ax.set_title('Average Course Progress by Month')
ax.set_xlabel('Month')
ax.set_ylabel('Average Course Progress (%)')
ax.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly_attendance.index.astype(str), monthly_attendance.values, marker='o', color='purple')
ax.set_title('Average Attendance Percentage by Month')
ax.set_xlabel('Month')
ax.set_ylabel('Average Attendance (%)')
ax.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


students_per_course = df.groupby('Course').size()
revenue_per_course = df.groupby('Course')['Revenue'].sum()
avg_score_per_course = df.groupby('Course')['FinalScore'].mean()
avg_attendance_per_course = df.groupby('Course')['AttendancePercentage'].mean()
students_per_city = df.groupby('City').size()
students_per_instructor = df.groupby('Instructor').size()

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(students_per_course.index, students_per_course.values, color='steelblue')
ax.set_title('Number of Students in Each Course')
ax.set_xlabel('Course')
ax.set_ylabel('Number of Students')
ax.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(revenue_per_course.index, revenue_per_course.values, color='green')
ax.set_title('Revenue Generated by Each Course')
ax.set_xlabel('Course')
ax.set_ylabel('Revenue')
ax.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(avg_score_per_course.index, avg_score_per_course.values, color='tomato')
ax.set_title('Average Final Score by Course')
ax.set_xlabel('Course')
ax.set_ylabel('Average Final Score')
ax.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(avg_attendance_per_course.index, avg_attendance_per_course.values, color='orange')
ax.set_title('Average Attendance Percentage by Course')
ax.set_xlabel('Course')
ax.set_ylabel('Average Attendance (%)')
ax.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(students_per_city.index, students_per_city.values, color='mediumpurple')
ax.set_title('Number of Students from Each City')
ax.set_xlabel('City')
ax.set_ylabel('Number of Students')
ax.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(students_per_instructor.index, students_per_instructor.values, color='teal')
ax.set_title('Number of Students Under Each Instructor')
ax.set_xlabel('Instructor')
ax.set_ylabel('Number of Students')
ax.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


top10_courses = df.groupby('Course').size().nlargest(10)
top10_instructors = df.groupby('Instructor').size().nlargest(10)
top10_students = df.nlargest(10, 'FinalScore')[['StudentName', 'FinalScore']].set_index('StudentName')

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top10_courses.index, top10_courses.values, color='steelblue')
ax.set_title('Top 10 Courses by Enrollment')
ax.set_xlabel('Number of Students')
ax.set_ylabel('Course')
ax.grid(axis='x')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top10_instructors.index, top10_instructors.values, color='darkorange')
ax.set_title('Top 10 Instructors by Student Enrollment')
ax.set_xlabel('Number of Students')
ax.set_ylabel('Instructor')
ax.grid(axis='x')
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top10_students.index, top10_students['FinalScore'].values, color='mediumseagreen')
ax.set_title('Top 10 Students with Highest Final Scores')
ax.set_xlabel('Final Score')
ax.set_ylabel('Student Name')
ax.grid(axis='x')
plt.tight_layout()
plt.show()


hist_configs = [
    ('CourseFee', 'Course Fee', 'Fee Amount', 'steelblue'),
    ('AttendancePercentage', 'Attendance Percentage', 'Attendance (%)', 'orange'),
    ('AssignmentScore', 'Assignment Score', 'Score', 'tomato'),
    ('QuizScore', 'Quiz Score', 'Score', 'mediumpurple'),
    ('FinalScore', 'Final Score', 'Score', 'green'),
    ('CourseProgress', 'Course Progress', 'Progress (%)', 'teal'),
]

for col, title, xlabel, color in hist_configs:
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.hist(df[col].dropna(), bins=20, color=color, edgecolor='black')
    ax.set_title(f'Distribution of {title}')
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Frequency')
    ax.grid(axis='y')
    plt.tight_layout()
    plt.show()


pie_configs = [
    ('Course', 'Student Distribution by Course'),
    ('Gender', 'Student Distribution by Gender'),
    ('PaymentMethod', 'Payment Method Distribution'),
    ('StudentStatus', 'Student Status Distribution'),
]

for col, title in pie_configs:
    counts = df[col].value_counts()
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=140)
    ax.set_title(title)
    plt.tight_layout()
    plt.show()

cert_counts = df['CertificateIssued'].value_counts()
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(cert_counts.values, labels=cert_counts.index, autopct='%1.1f%%', startangle=140, colors=['mediumseagreen', 'tomato'])
ax.set_title('Certificate Issued vs Not Issued')
plt.tight_layout()
plt.show()


scatter_configs = [
    ('AttendancePercentage', 'FinalScore', 'Attendance Percentage', 'Final Score', 'steelblue'),
    ('AssignmentScore', 'FinalScore', 'Assignment Score', 'Final Score', 'orange'),
    ('QuizScore', 'FinalScore', 'Quiz Score', 'Final Score', 'tomato'),
    ('StudyHoursPerWeek', 'FinalScore', 'Study Hours Per Week', 'Final Score', 'green'),
    ('CourseProgress', 'FinalScore', 'Course Progress', 'Final Score', 'mediumpurple'),
    ('AttendancePercentage', 'CourseProgress', 'Attendance Percentage', 'Course Progress', 'teal'),
]

for x_col, y_col, xlabel, ylabel, color in scatter_configs:
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.scatter(df[x_col], df[y_col], alpha=0.5, color=color, edgecolors='none')
    ax.set_title(f'{xlabel} vs {ylabel}')
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.grid(True)
    plt.tight_layout()
    plt.show()
```
