#Numpy

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Datasets/daraz_dataset.csv")
df.head()
df.shape
df.columns
df.dtypes
df['Age']
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors="coerce")
```

```python
df.dtypes
```

```python
numeric_columns = [
    "Age",
    "Quantity",
    "UnitPrice",
    "DiscountPercent",
    "TotalAmount",
    "Rating"
]
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean)
df.isnull().sum()
```

```python
fearures = ["Age", "Quantity", "UnitPrice", "DiscountPercent"]
target = "TotalAmount"
X = df[fearures]
y = df[target]

X = X.to_numpy()
y = y.to_numpy()
```
